```python
import pandas as pandas
import requests
from bs4 import BeautifulSoup
from openai import OpenAI
import os
from dotenv import load_dotenv
```

```python
PATH='temp/urls.txt'
urls = []
with open(PATH, 'r') as file:
    urls = file.read().splitlines()
```

```python
url = urls[0]
response = requests.get(url)
soup = BeautifulSoup(response.content, 'html.parser')
href = [link.get('href') for link in soup.find_all('a')]
admission_links = []
for link in href:
    if 'admission' in link:
        admission_links.append(link)
```

```python
load_dotenv()
api_key = os.getenv('GROQ_API')

client = OpenAI(
    api_key=api_key,
    base_url="https://api.groq.com/openai/v1"
)
```

```python
# Create a completion request
completion = client.chat.completions.create(
    model="llama-3.3-70b-versatile", # You can use other models like 'mixtral-8x7b-32768'
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "State the Country, city, state and postal code of the university in the following link: " + url}
    ]
)

# Extract and print the answer
print(completion.choices[0].message.content)
```
