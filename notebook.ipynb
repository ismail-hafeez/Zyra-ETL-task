```python
load_dotenv()
api_key = os.getenv('GROQ_API')

client = OpenAI(
    api_key=api_key,
    base_url="https://api.groq.com/openai/v1"
)
```

```python
query = f"""
State the Country, city, state and postal code of the university in the following link: {urls[0]}
Return ONLY the country, city, state and postal code.
"""
```

```python
# Create a completion request
completion = client.chat.completions.create(
    model="llama-3.3-70b-versatile", # You can use other models like 'mixtral-8x7b-32768'
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": query}
    ]
)

# Extract and print the answer
ans=completion.choices[0].message.content
print(completion.choices[0].message.content)
```

```python
type(ans)
```

```python
ans=ans.strip().split(", ")
```

```python
country = ans[0]
city = ans[1]
state = ans[2]
postal_code = ans[3]
```

```python
print(f"Country: {country}")
print(f"City: {city}")
print(f"State: {state}")
print(f"Postal Code: {postal_code}")
```

```python
PATH='./data/university_domains.txt'

def get_domain() -> Generator[str, None, None]:
    with open(PATH, 'r') as f:
        urls = f.read().splitlines()
    for url in urls:
        yield url
```

```python
domain_gen = get_domain()

while True:
    try:
        url = next(domain_gen)
        print(f"Processing: {url}")
        # Your logic here
    except StopIteration:
        # This triggers when the generator runs out of items
        print("Finished processing all domains.")
        break
```
